# Monte Carlo
## Sampling from multiple distributions to do numerical intergration

Install MASS

```r
#Loading libraries
req_packages<-c("MASS")

for(i in c(1:length(req_packages))){
  if (!require(req_packages[i], character.only = TRUE)){
   install.packages(req_packages[i])
  }
}
```

# Example 1

```r
library(MASS)
n=10000

V=matrix(0,2,2)
V[1,1]=10
V[2,2]=1
random=mvrnorm(n,c(3,5),V)
#vector to store variable 1 - variable 2
s=rep(0,n)
i=1
#track the count of the number of times variable 1 > variable 2
probmc=0
while(i<=n){

	s[i]=random[i,1]-random[i,2]
	#count the number of times variable 1 is greater variable 2
	if(random[i,1]>random[i,2]){
			probmc=probmc+1
	}
	i=i+1
}

probmc=probmc/n
print(probmc)
```
